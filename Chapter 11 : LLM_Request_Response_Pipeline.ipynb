| Method | Async? | Requests | Response type | Use case |
|---|---|---|---|---|
| `invoke()` | ❌ No | One | Complete response | One LLM request |
| `ainvoke()` | ✅ Yes | One | Complete response | One async LLM request |
| `batch()` | ❌ No | Many | Complete responses | Multiple independent requests |
| `abatch()` | ✅ Yes | Many | Complete responses | Multiple independent async requests |
| `stream()` | ❌ No | One | Response chunks | Show response progressively |
| `astream()` | ✅ Yes | One | Response chunks | Async progressive response |

```python
├── invoke()       → single request
├── ainvoke()      → async request
├── batch()        → multiple requests
├── abatch()       → async batch
├── stream()       → streaming response
└── astream()      → async streaming
```

```python
 # 4) abatch -> async batch

 # Use abatch() when you have multiple independent LLM requests and you don't need to wait for one request before starting the next.
```

```python
prompts = [
    "Explain RAG in simple words.",
    "Explain embeddings in simple words.",
    "Explain vector databases in simple words."
]

responses4 = await llm.abatch(prompts)

print(responses4)
```

```python
for i in responses4:
    print(i.content)
```

```python
for i in responses4:
    print(i.response_metadata)
```

```python
5) stream
```

```python
prompt = "Explain RAG in simple words"

response5 = llm.stream(prompt)
```

```python
for chunk in response5:
    print(chunk)
```

```python
for chunk in response5:
    print(chunk.content, end="", flush=True)
```

```python
response6 = llm.astream("Explain RAG in simple words.")

async for chunk in response6:
    print(chunk.content, end="", flush=True)
```
