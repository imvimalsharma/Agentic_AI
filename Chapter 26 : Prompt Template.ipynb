# Chapter 26 : Prompt Template

A prompt template is a reusable, structured pattern (or “blueprint”) for writing prompts that you send to a large language model (LLM).Instead of hard-coding a full prompt every time, you define a template that contains:Fixed instructions / role / style
Placeholders (variables) that get filled in with actual data at runtime



```python
from dotenv import load_dotenv
load_dotenv()
import os

groq_api_key = os.getenv("GROQ_API_KEY")
```

```python
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(model="openai/gpt-oss-20b",
    base_url="https://api.groq.com/openai/v1",
    api_key=os.environ["GROQ_API_KEY"])
```

```python
from langchain_core.prompts import PromptTemplate

response = PromptTemplate.from_template(
    "Explain {topic} to a {level}"
)

prompt = response.format(
    topic = input("Tell us about topic :"),
    level = input("Beginner, Experienced")
)

print(llm.invoke(prompt).content)
```

The LCEL (LangChain Expression Language) approach is a declarative way to build and connect AI components—like prompts, 
language models, output parsers, and retrievers—into unified workflows using the native pipe operator (|).

```python
from langchain_core.prompts import PromptTemplate

prompt_template = PromptTemplate.from_template(
    """
You are an AI learning assistant.

Topic: {topic}
Difficulty: {level}
Output format: {format_type}

Instructions:
1. Explain the topic at the requested level.
2. Include one practical example.
3. Keep the explanation focused.
"""
)

chain = prompt_template | llm

try:
    response = chain.invoke({
        "topic": input("Tell us your topic"),
        "level": input("Tell us the experience level"),
        "format_type": input("Tell us how you want output")
    })

    print(response.content)

except Exception as e:
    print("Request failed:", e)
```
