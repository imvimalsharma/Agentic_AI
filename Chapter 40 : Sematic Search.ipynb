# Chapter 39 — Build a Semantic Search Engine
Goal: Build a search engine that retrieves relevant sentences based on meaning, not just exact keyword matches.
In your previous chapters, you generated embeddings, calculated cosine similarity, compared embedding models, and evaluated their quality. Now we'll combine those skills into a working application.

How semantic search works
Document collection
Store sentences or documents

Generate embeddings
Convert documents into vectors

Embed the user's query
Use the same embedding model

Calculate cosine similarity
Rank documents by similarity

Return top results

# Step 1: Create a document

```python
from sentence_transformers import SentenceTransformer

model_small = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")
```

```python
documents = [
    "Banks monitor international payments to detect suspicious transactions.",
    "Money laundering involves concealing the origin of illegally obtained funds.",
    "Machine learning models can identify unusual transaction patterns.",
    "Credit risk analysis estimates the likelihood of borrowers defaulting.",
    "Cross-border payments transfer money between different countries.",
    "Customer due diligence helps banks understand their customers.",
    "Neural networks are used for image recognition and language tasks.",
    "Transaction monitoring systems generate alerts for potentially suspicious activity."
]
```

```python
# Cosine Similarity as discussed before 
```

```python
# Cosine_ imilarity Function 
import numpy as np 

def cosine_similarity(vector_a, vector_b):
    dot_product = np.dot(vector_a, vector_b)

    mag_a = np.linalg.norm(vector_a)
    mag_b = np.linalg.norm(vector_b)

    if mag_a == 0 or mag_b == 0:
        raise ValueError("Cannot compare a zero vector")

    return float(dot_product/(mag_a*mag_b))


```

```python
# Step 2: Generate Document Embeddings

document_embeddings = model_small.encode(documents)
```

```python
print("Number of Documents:" , len(document_embeddings))
```

```python
print("Embedding matrix shape:", document_embeddings.shape)
```

# Step 3 : Build the search functions

```python
def semantic_search(query, documents, document_embeddings, top_k = 3 ):
    
    query_embedding = model_small.encode([query])[0]

    scores = []

    for i, document_vector in enumerate(document_embeddings):
        score = cosine_similarity(query_embedding, document_vector)
        scores.append({
            "documents" : documents[i],
            "score" : score
        })
        ranked_results = sorted(scores, key = lambda item: item["score"], reverse=True)

        return ranked_results[:top_k]
```

model_small.encode([query])[0] converts the query into one vector.

The for loop compares that vector with every document vector.

scores.append(...) stores each document and its score.



sorted(..., reverse=True) puts the highest scores first.



[:top_k] returns only the requested number of results.

```python
query = "How do banks detect suspicious international payments ? "

results = semantic_search(query, documents, document_embeddings, top_k = 3)
```

```python
results
```

```python
for rank, result in enumerate(results, start=1):
    print(f"\nRank: {rank}")
    print(f"Similarity: {result['score']:.4f}")
    print(f"Document: {result['documents']}")
```

What you've built
You now have a basic semantic search engine with:
- Document embedding and storage in memory.
- Query embedding.
- Cosine similarity ranking.
- Top-\(k\) result retrieval.

One limitation: this version compares the query against every document, so its search cost grows with the size of the collection. It also has no persistent storage yet.
