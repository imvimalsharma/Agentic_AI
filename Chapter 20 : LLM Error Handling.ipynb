````python
from langchain_openai import ChatOpenAI
import os

llm = ChatOpenAI(
    model="openai/gpt-oss-20b",
    base_url="https://api.groq.com/openai/v1",
    api_key=os.environ["GROQ_API_KEY"],

)
````

````python
prompt = "What is RAG"
````

````python
def ask_llm(prompt):

  try:
    response = llm.invoke(prompt)

    return response.content

  except Exception as e:
    print("LLM has failed")
    print("Error", e)

    return None
````

````python
answer = ask_llm("Explain Retrieval Augmented Generation in simple words.")
print(answer)
````

# Now will add retry

````python
import time

def ask_llm(prompt, max_retries = 3):
  for attempt in range(max_retries):

    try:
      response = llm.invoke(prompt)
      return response.content

    except Exception as e:

       print(f"Attempt {attempt + 1} failed.")
       print("Error:", e)

       if attempt < max_retries -1:
            print("retrying")
            time.sleep(2)

       else:
          print("All retries failed.")
          return None

````

````python
ask_llm("Give me picture of Salman khan ")
````
