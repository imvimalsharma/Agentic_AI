# Chapter 24 — Prompt Variables: Dynamic Prompt Generator

# What are prompt variables?
# Suppose you want the LLM to explain different topics at different difficulty levels.
# Without variables, you might write separate prompts:
# llm.invoke("Explain RAG to a beginner.")
# llm.invoke("Explain embeddings to an expert.")
# llm.invoke("Explain vector databases to a beginner.")


# With variables, you create one reusable prompt:
# prompt = f"Explain {topic} to a {level}."

# Now topic and level can change dynamically.

```python
from dotenv import load_dotenv
load_dotenv()
import os

groq_api_key = os.getenv("GROQ_API_KEY")
```

```python
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(model="openai/gpt-oss-20b",
    base_url="https://api.groq.com/openai/v1",
    api_key=os.environ["GROQ_API_KEY"])
```

```python
def dynamic_prompt(topic, level, format_type):
    prompt = f"""
Explain {topic} to a {level}.

Requirements:
- Use appropriate language for the audience.
- Format the answer as {format_type}.
- Include a practical example.
"""
    return prompt


topic = input("Enter your Topic :" )
level = input("Enter you difficulty level of topic :")
format_type = input("Enter the style you want ")

prompt = dynamic_prompt(topic, level, format_type)

response = llm.invoke(prompt)

print(response.content)
```
