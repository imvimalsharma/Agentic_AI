
# Chapter 19: Streaming Chatbot




With a normal chatbot, you send the question and wait for the complete response.

```python
## Example
```

*You*: Explain RAG.

          Waiting...

AI: RAG stands for Retrieval Augmented Generation.
    It allows an LLM to retrieve relevant information
    from external documents before generating an answer.

A streaming chatbot doesn't wait for the entire response.
The LLM generates the response progressively, and your application displays each piece as it arrives.

```python
# To the use it looks like AI is typing
```

| | Normal | Streaming |
|---|---|---|
| Method | `invoke()` | `stream()` |
| Response | Complete response | Chunks |
| Display | At the end | As generated |
| User sees typing effect | ❌ | ✅ |
| Better perceived responsiveness | ❌ | ✅ |
| Final answer | Same | Same |
| Main difference | **Delivery** | **Delivery** |

```python
from langchain_openai import ChatOpenAI
import os

llm = ChatOpenAI(
    model="openai/gpt-oss-20b",
    base_url="https://api.groq.com/openai/v1",
    api_key=os.environ["GROQ_API_KEY"],

)
```

```python
prompt = " What is RAG"
```

```python
response = llm.stream(prompt)
```

```python
for chunk in response:
    print(chunk.content, end="", flush = True)
```

```python
# Lets make now streaming chatbot
```

```python
from langchain_core.messages import (SystemMessage,HumanMessage,AIMessage)

messages = [SystemMessage(content="You are a helpful AI assistant."]

while True:

    user_input = input("\nYou: ")

    if user_input.lower() == "exit":
        print("Chat ended.")
        break

    # 1. Add user message to history
    messages.append(HumanMessage(content=user_input))

    print("\nAI: ", end="", flush=True)

    # 2. Stream using the COMPLETE history
    full_response = ""

    for chunk in llm.stream(messages):

        if chunk.content:
            print(chunk.content, end="", flush=True)

            # Store each chunk
            full_response += chunk.content

    print()

    # 3. Add complete AI response to history
    messages.append(AIMessage(content=full_response))
```
