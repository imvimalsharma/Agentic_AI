In Chapter 34, you learned cosine similarity, which measures the angle between two vectors. Now, in Chapter 35, we'll learn Euclidean distance, which measures the straight-line distance between two vectors.

```python
# Calculate distance manually 
import numpy as np

vector_a = np.array([1,2])
vector_b = np.array([4,6])

difference = vector_a - vector_b

squared_difference = difference ** 2

sum_squared = np.sum(squared_difference)

distance = np.sqrt(sum_squared)
```

# Now we will build a resuable function 

```python
def euclidean_distance(vector_a, vector_b):
    difference = vector_a - vector_b
    distance = np.linalg.norm(difference)
    return float(distance)
```

```python
a = np.array([1, 2])
b = np.array([4, 6])

print(euclidean_distance(a, b))
```

```python
# Compare cosine vs euclidean_distance similarity
```

```python
def cosine_similarity(vector_a, vector_b):
    dot_product = np.dot(vector_a, vector_b)

    magnitiude_a = np.linalg.norm(vector_a)
    magnitiude_b = np.linalg.norm(vector_b)

    if magnitiude_a == 0 or magnitiude_b == 0:
        raise ValueError("Cannot compare a zero vector")

    similarity = dot_product/(magnitiude_a * magnitiude_b)
    return float(similarity)
```

```python
a = np.array([1, 2])
b = np.array([2, 4])
c = np.array([4, 2])

print("A vs B")
print("Cosine:", cosine_similarity(a, b))
print("Distance:", euclidean_distance(a, b))

print("\nA vs C")
print("Cosine:", cosine_similarity(a, c))
print("Distance:", euclidean_distance(a, c))
```
