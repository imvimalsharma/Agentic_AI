# What is Few-Shot Prompting?
Few-shot prompting means providing a small number of examples in the prompt before asking the LLM to perform a task.

```python
from dotenv import load_dotenv
load_dotenv()
import os 

groq_api_key = os.getenv("GROQ_API_KEY")

if not groq_api_key:
    raise ValueError("API Key Not Found")
```

```python
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(    model="openai/gpt-oss-20b",
    base_url="https://api.groq.com/openai/v1",
    api_key=os.environ["GROQ_API_KEY"],)
```

```python
# We are avoiding logs because at this stage it is not required,will do it later to simplify
```

```python
import time
def few_shot_text(text):
    examples = """
Example 1:
Text: I absolutely love this product.
Sentiment: Positive

Example 2:
Text: The delivery was late and the quality was terrible.
Sentiment: Negative

Example 3:
Text: The package arrived this morning.
Sentiment: Neutral

Example 4:
Text: Customer support solved my problem quickly.
Sentiment: Positive

Example 5:
Text: I am disappointed with the service.
Sentiment: Negative
"""
    categories = ["Positive", "Negative","Neutral"]
    prompt = f"""
You are a sentiment classification system.

Use the examples below to understand the task.

{examples}

Classify the following new text into exactly ONE category:
{categories}

Rules:
1. Return only Positive, Negative, or Neutral.
2. Do not provide an explanation.
3. Do not invent another category.

Text: {text}

Sentiment:
"""

    start_time = time.time()
    try:
        response = llm.invoke(prompt)
        latency = round(time.time() - start_time,3)
        result = response.content.strip()
        if result not in categories:
            raise ValueError(f"Result not found in {categories}")
        return result
    except Exception as e : 
        print("Classification Failed", e)

        return None
```

```python
text = "The application is easy to use and works perfectly."
result = few_shot_text(text)
```
