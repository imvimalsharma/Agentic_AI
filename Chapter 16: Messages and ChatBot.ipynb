# Chapter 16: Different types of messages




| Message | Represents | Purpose |
|---|---|---|
| `SystemMessage` | System/application | Instructions |
| `HumanMessage` | User | User input |
| `AIMessage` | AI | Previous AI response |
| `ToolMessage` | Tool | Result returned by a tool |

```python
from langchain_openai import ChatOpenAI
import os

llm = ChatOpenAI(
    model="openai/gpt-oss-20b",
    base_url="https://api.groq.com/openai/v1",
    api_key=os.environ["GROQ_API_KEY"],

)
```

```python
# SYSTEM MESSAGE :This contains instructions for the AI. Its like how AI should behave.
```

```python
from langchain_core.messages import SystemMessage

message = SystemMessage(
    content="""You are a financial crime analyst.
    Explain concepts in simple language.
    Always provide examples.
    """
)
```

```python
# Human Message : This represents the user's message.
```

```python
from langchain_core.messages import HumanMessage

message = HumanMessage(
    content="What is RAG?"
)
```

```python
print(type(message))
```

```python
# AI message : LLM whcih send this message
```

```python
response = llm.invoke([message])
```

```python
# Now this proves its an AI message
```

```python
print(type(response))
```

```python
# Lets make a simple chatbot with Memory
```

```python
from langchain_core.messages import (SystemMessage, HumanMessage)

messages = [SystemMessage(content="You are a helpful AI assistant.")]

while True:

    user_input = input("\nYou: ")

    if user_input.lower() == "exit":
        break

    messages.append(
        HumanMessage(content=user_input)
    )

    response = llm.invoke(messages)

    messages.append(response)

    print("\nAI:", response.content)
```
