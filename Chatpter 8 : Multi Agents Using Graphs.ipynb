| Type                  | Simple definition                                                       | Example                                                                          |
| --------------------- | ----------------------------------------------------------------------- | -------------------------------------------------------------------------------- |
| **Context Memory**    | Information available to the AI **right now** for the current response. | User says, “My name is Vimal.” → AI can use “Vimal” in the current conversation. |
| **Short-Term Memory** | Information the AI keeps **during the current task/session**.           | Researcher finds data → Analyst uses that data later in the same workflow.       |
| **Long-Term Memory**  | Information **stored for future conversations or tasks**.               | User says, “I prefer Python.” → System stores this preference and uses it later. |
| **Knowledge Memory**  | External information the AI can **retrieve when needed**.               | AI retrieves information about AML rules from a company policy document.         |

# Conversational Buffer Memory

Conversational Buffer Memory is a type of memory where the AI stores the conversation history as a sequence of messages and sends that history back to the LLM when generating the next response.

```python
from langchain.agents import create_agent

agent = create_agent(
    model=llm,
    system_prompt="You are a helpful assistant.",
    checkpointer=memory
)
```

```python
messages = []

while True:

    question = input("You: ")

    if question.strip().lower() == "exit":
        print("Conversation ended.")
        break

    messages.append({
        "role": "user",
        "content": question
    })

    response = llm.invoke(messages)

    messages.append({
        "role": "assistant",
        "content": response.content
    })

    print("AI:", response.content)
```
