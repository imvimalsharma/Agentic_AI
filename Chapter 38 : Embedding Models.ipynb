# Chapter 38 : Embedding Models

So far, you've used sentence-transformers/all-MiniLM-L6-v2, which produces 384-dimensional vectors.
But different embedding models can represent the same sentence differently. We'll compare two models and investigate their vector dimensions, similarity scores, and performance.

```python
from sentence_transformers import SentenceTransformer

model_small = SentenceTransformer(
    "sentence-transformers/all-MiniLM-L6-v2"
)

model_large = SentenceTransformer(
    "sentence-transformers/all-mpnet-base-v2"
)
```

all-MiniLM-L6-v2: produces 384-dimensional embeddings and is relatively lightweight.



all-mpnet-base-v2: produces 768-dimensional embeddings and may capture semantic relationships differently, at a higher computational cost.

```python
sentences = [
    "The bank approved my loan application.",
    "My loan request was accepted by the bank.",
    "The cricket team won the match."
]
```

```python
embeddings_small = model_small.encode(sentences)
embeddings_large = model_large.encode(sentences)
```

```python

print("Small model shape:", embeddings_small.shape)
print("Large model shape:", embeddings_large.shape)
```

```python
# Checking the Similarity Score using cosine Similarity 
```

```python
import numpy as np
```

```python
def cosine_similarity(vector_a, vector_b):
    dot_product = np.dot(vector_a, vector_b)

    magnitiude_a = np.linalg.norm(vector_a)
    magnitiude_b = np.linalg.norm(vector_b)

    if magnitiude_a == 0 or magnitiude_b == 0:
        raise ValueError("Cannot compare a zero vector")

    similarity = dot_product/(magnitiude_a * magnitiude_b)
    return float(similarity)
```

```python
for name, embeddings in [("MiniLm", embeddings_small),
                         ("MPNet", embeddings_large)] :
    score_related = cosine_similarity(embeddings[0],embeddings[1])

    score_unreleated =cosine_similarity(embeddings[0],embeddings[2])

    print(f"Model :{name}")
    print("Related sentences",round(score_related,4))
    print("Different topics", round(score_unreleated,4))
```

```python
#Benchmarking encoding time 
import time

for name, embeddings_model in [("MiniLm", model_small),
                         ("MPNet", model_large)] :
    start_time = time.perf_counter()
    vectors = embeddings_model.encode(sentences)
    elapsed = time.perf_counter() - start_time
    print(f"{name}")
    print("Dimension", vectors.shape[1])
    print("encoding time", round(elapsed, 4),"seconds")
```
