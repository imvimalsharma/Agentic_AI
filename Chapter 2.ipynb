# Chapter 2 : Prompt Chaining
What is Prompt Chaining ?

Output of Prompt 1 becomes input of Prompt 2

# Doing API Authentication

```python
from google import genai
from google.genai import types

# Developer TODO: Replace YOUR_API_KEY with your API key.
API_KEY = "Please enter your API key"

client = genai.Client(
    vertexai=False, api_key=API_KEY
)
```

```python
import os

os.environ["GEMINI_API_KEY"] = "Please enter your API key"
```

```python
chat = client.chats.create(model="gemini-3.5-flash")
```

```python
# Step 1: Getting the first prompt as input
```

```python
response_2 = chat.send_message("Give me 5 common AML transaction patterns. Give me the transaction and types in table format")
```

```python
print(response_2.text)
```

```python
# Step 2 : Taking the input of Prompt 1 as input of Prompt 2
```

```python
response_3 = chat.send_message(f"{response_2.text} Group these patterns into broader categories.")
```

```python
print(response_3.text)
```

```python
# For testing: We can get the results in JSON Format
```

```python
response_4 = chat.send_message(
    """Give me 5 common AML transaction patterns.

    Return the answer ONLY as valid JSON in this format:

    {
        "patterns": [
            {
                "name": "pattern name",
                "description": "short description"
            }
        ]
    }
    """
)
```

```python
print(response_4.text)
```
