#  What are embeddings?
An embedding converts text into a list of numbers called a vector.

```python
# "Apple released a new AI chip"
#               ↓
#       Embedding Model
#               ↓
# [0.12, -0.34, 0.87, 0.21, ...]
```

These numbers represent aspects of the text's learned meaning. Texts with similar meanings often have vectors that are close together in vector space.
An embedding is not a summary or a classification label. It is a numerical representation useful for comparing and retrieving information.

# Embedding model
Example: sentence-transformers/all-MiniLM-L6-v2

Converts text into numerical vectors for similarity search.

For this chapter, we'll use a local embedding model so you can generate vectors without making another paid API request.

```python
# Install it 
%pip install -U sentence-transformers
```

```python
from sentence_transformers import SentenceTransformer
```

```python
model = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")

text = "Artificial intelligence helps computers understand language."

embedding = model.encode(text)

print(text)
```

```python
print("Embedding:", embedding)
```

```python
print("Vector dimensions:", embedding.shape)
```

The model generates a vector containing 384 numerical values for each input text. Your first run downloads the model, so it requires internet access; subsequent runs can use the locally cached model.

```python
texts = [
    "Artificial intelligence helps computers understand language.",
    "Machine learning enables computers to learn from data.",
    "The Indian cricket team won the match.",
    "Vector databases store numerical representations of data."
]
```

```python
embeddings = model.encode(texts)
```

```python
print("Number of texts:", len(texts))
```

```python
print("Embedding matrix shape:", embeddings.shape)
```

```python
# Inspect the vectors
# Let's examine the first five dimensions rather than printing every number.
```

```python
for text, vector in zip(texts, embeddings):
    print(text)
    print(vector[:5])
    print("_"*40)
```

Your embedding model converts each sentence into 384 numbers. These numbers represent the sentence in a high-dimensional space, which we cannot directly visualize.
