#  What is role-based prompting?
Role-based prompting means instructing the LLM to behave as a particular type of assistant.
For example:
- A Python tutor explains code to beginners.
- A financial analyst examines business performance.
- An AML analyst reviews suspicious transaction patterns.
- A technical interviewer asks questions and evaluates answers.
The model is the same. The instructions change how it approaches the task.



2. How it works

Role

You are a financial crime analyst.

Task

Review a transaction description.

Constraints

Explain risk indicators and uncertainty.

Role-aligned response

```python
from dotenv import load_dotenv
load_dotenv()
import os

groq_api_key = os.getenv("GROQ_API_KEY")
```

```python
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(model="openai/gpt-oss-20b",
    base_url="https://api.groq.com/openai/v1",
    api_key=os.environ["GROQ_API_KEY"])
```

```python
from langchain_core.messages import SystemMessage, HumanMessage


def role_assistant(role, task):

    system_message = SystemMessage(
        content=f"""
You are {role}.

Follow the responsibilities and communication
style appropriate for your role.

Be accurate, clear, and acknowledge uncertainty.
"""
    )

    user_message = HumanMessage(content=task)

    try:
        response = llm.invoke([
            system_message,
            user_message
        ])

        return response.content

    except Exception as e:
        print("Request failed:", e)
        return None


# Test
role = "a senior Python engineer"
task = "Explain why Python functions are useful."

result = role_assistant(role, task)

print(result)
```
