# Chapter 37 : Semantic Similarity Checker

Goal: Build a Python program that accepts two sentences and determines how semantically similar they are using embeddings and cosine similarity.

```python
import numpy as np
from sentence_transformers import SentenceTransformer

model = SentenceTransformer(
    "sentence-transformers/all-MiniLM-L6-v2"
)
```

```python
def cosine_similarity(vector_a, vector_b):
    dot_product = np.dot(vector_a, vector_b)

    magnitiude_a = np.linalg.norm(vector_a)
    magnitiude_b = np.linalg.norm(vector_b)

    if magnitiude_a == 0 or magnitiude_b == 0:
        raise ValueError("Cannot compare a zero vector")

    similarity = dot_product/(magnitiude_a * magnitiude_b)
    return float(similarity)
```

```python
def check_similarity(sentence_a, sentence_b):
    # Convert both sentences into embeddings

    embeddings = model.encode([sentence_a, sentence_b])

    vector_a = embeddings[0]
    vector_b = embeddings[1]

    # calculate cosine_similarity
    score = cosine_similarity(vector_a, vector_b)

    # Display results
    print("Sentence A:", sentence_a)
    print("Sentence B:", sentence_b)
    print("Similarity score:", round(score, 4))

    return score
```

Understand the important lines
- model.encode([sentence_a, sentence_b]) generates two vectors in one call.
- embeddings[0] selects the first sentence's vector.
- embeddings[1] selects the second sentence's vector.
- cosine_similarity(vector_a, vector_b) calculates their similarity.
- round(score, 4) rounds the displayed result to four decimal places.
- return score allows you to reuse the numerical result elsewhere in your program.

```python
check_similarity(
    "The bank approved my loan.",
    "My loan application was accepted by the bank."
)
```

```python
# Now two unrelated sentences
check_similarity(
    "The bank approved my loan.",
    "The football team won the match."
)
```

# Now we will add similarity 

```python
def check_similarity_label(sentence_a, sentence_b):
    embeddings = model.encode([sentence_a, sentence_b])

    score = cosine_similarity(embeddings[0], embeddings[1])

    if score >= 0.70:
        label = "High Similarity"
    elif score >= 0.40:
        label = "Moderately similar"
    else: 
        label = "Less Similar "

    return{round(score, 4), label}
```

```python
check_similarity_label(
    "The bank approved my loan.",
    "My loan application was accepted by the bank."
)
```

```python
# Now two unrelated sentences
check_similarity_label(
    "The bank approved my loan.",
    "The football team won the match."
)
```
