# Theory

An LLM doesn't directly produce a complete sentence at once.
It generates one token at a time.

Before probabilities are calculated, the model produces logits.

Temperature controls how concentrated or spread out the probability distribution is during sampling.

When Low Temperature the probability distribution becomes more concentrated.

Low temperature → more predictable output

Running the same prompt multiple times is more likely to produce similar responses.

Temperature
Controls probability distribution sharpness.
Top-P
Limits sampling to a group of tokens whose cumulative probability reaches a specified threshold.
Top-K
Limits sampling to the top K most probable tokens.

```python
from langchain_openai import ChatOpenAI
import os

llm = ChatOpenAI(
    model="openai/gpt-oss-20b",
    base_url="https://api.groq.com/openai/v1",
    api_key=os.environ["GROQ_API_KEY"],

)
```

```python
model = "openai/gpt-oss-20b"
```

```python
prompt = " Give me a creative name for an AI research assistant"
```

```python
temperatures = [ 0.0, 0.1, 0.5]
```

bind() is a LangChain method that creates a new version of your model with specific parameters attached to it.

```python
for temperature in temperatures:
  model = llm.bind(temperature = temperature)

  response = model.invoke(prompt)

  print("TEMPERATURE:", temperature)
  print(response.content)
```

# Checking the tokens usage for different temperature

```python
results = []

prompt = "Explain RAG in three sentences."

temperatures = [0.1, 0.5, 0.9]

for temperature in temperatures:

    model = llm.bind(temperature=temperature)

    for run in range(5):

        response = model.invoke(prompt)

        usage = response.usage_metadata

        results.append({
            "Temperature": temperature,
            "Run": run + 1,
            "Response": response.content,
            "Input Tokens": usage.get("input_tokens"),
            "Output Tokens": usage.get("output_tokens"),
            "Total Tokens": usage.get("total_tokens")
        })
```

```python
import pandas as pd

df = pd.DataFrame(results)

df
```
