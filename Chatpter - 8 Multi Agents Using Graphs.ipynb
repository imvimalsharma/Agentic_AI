```python
from langchain_openai import ChatOpenAI
import os

llm = ChatOpenAI(
    model="openai/gpt-oss-20b",
    base_url="https://api.groq.com/openai/v1",
    api_key=os.environ["GROQ_API_KEY"]
)
```

```python
model = "openai/gpt-oss-20b"
```

```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
```

```python
class State(TypedDict):
    question: str
    research: str
    analysis: str
    final_answer: str
```

```python
# Research Agents

def researcher(state : State) :
      response = llm.invoke([
        {
            "role": "system",
            "content": """
            You are a Senior Research Agent.

            Research the user's question carefully.
            Identify the important facts and information
            required to answer the question.
            """
        },
        {
            "role": "user",
            "content": state["question"]
        }
    ])

      return {
        "research": response.content
      }
```

```python
def analyst(state: State):

    response = llm.invoke([
        {
            "role": "system",
            "content": """
            You are a Senior Analyst.

            Analyze the research provided by the researcher.
            Identify the key insights, trends, opportunities,
            risks and important conclusions.
            """
        },
        {
            "role": "user",
            "content": f"""
            Original Question:
            {state["question"]}

            Research:
            {state["research"]}
            """
        }
    ])

    return {
        "analysis": response.content
    }
```

```python
def writer(state: State):

    response = llm.invoke([
        {
            "role": "system",
            "content": """
            You are a Senior Technical Writer.

            Create a clear, structured final answer
            using the research and analysis provided.
            """
        },
        {
            "role": "user",
            "content": f"""
            Question:
            {state["question"]}

            Research:
            {state["research"]}

            Analysis:
            {state["analysis"]}

            Write the final answer.
            """
        }
    ])

    return {
        "final_answer": response.content
    }
```

```python
# We will Build a graph
```

```python
from langgraph.graph import StateGraph, START, END
```

```python
graph = StateGraph(State)
graph.add_node("researcher", researcher)
graph.add_node("analyst", analyst)
graph.add_node("writer", writer)
```

```python

graph.add_edge(START, "researcher")
graph.add_edge("researcher", "analyst")
graph.add_edge("analyst", "writer")
graph.add_edge("writer", END)
```

```python
app = graph.compile()
```

```python
question = input("Enter the topic you want to research: ")
```

```python
# 9. Run Multi-Agent System
# ==========================================

result = app.invoke({
    "question": question,
    "research": "",
    "analysis": "",
    "final_answer": ""
})
```
